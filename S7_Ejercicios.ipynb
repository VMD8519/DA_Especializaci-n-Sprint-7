Escribe una función que calcule el IMC ingresado por el usuario/a, es decir, quien lo ejecute deberá ingresar estos datos. Puedes obtener más información de su cálculo en:

-> Índice de masa corporal IMC que es y cómo se calcula.

La función debe clasificar el resultado en sus respectivas categorías

PAra realizar este ejercicio, cree las siguientes funciones que las uní luego en una única funcion llamada "main". A continuacion detallo las funciones: 

#FUNCION 1-> Solicitud de datos
Lo primero que hice fue diseñar una función llamada "Solicitud_datos" que utiliza el comando input para permitir al usuario ingresar su altura en metros y su peso en kilogramos. Esta función incluye mensajes para guiar al usuario durante la entrada de los datos, asegurando que comprendan qué tipo de valores deben proporcionar.
Ambas entradas (altura y peso) se convierten a valores de tipo float para garantizar que puedan ser utilizados en cálculos matemáticos posteriores. Finalmente, la función devuelve tanto la altura como el peso utilizando return, lo que permite que estos valores sean utilizados en otras partes del código.

#FUNCION 2 -> Calculo IMC
La fórmula para calcularlo es:IMC = Peso / Altura². 
Para su calculo, hice una función llamada "calculo_IMC" que toma como parámetros de entrada la altura y el peso proporcionados por la función de "Solicitud_datos". Dentro de la función, se realiza el cálculo matemático necesario para determinar el IMC. El resultado se devuelve utilizando la instrucción return.

#FUNCION 3 -> Validaciones
Esta función se encarga de realizar la validación de la clasificación del IMC dentro de las categorías definidas por la OMS. Según el valor del IMC calculado, se asigna una categoría específica, como "Bajo peso", "Peso normal", "Sobrepeso","Obesidad", etc. Finalmente, la categoría correspondiente se devuelve utilizando la instrucción return.

#FUNCIÓN GLOBAL -> MAIN
Esta función agrupa las tres funciones previamente descritas, integrándolas en un único flujo de trabajo para facilitar su ejecución. Al unificar las funciones de solicitud de datos, cálculo del IMC y validaciones, la función MAIN actúa como la función central. Posteriormente, esta función se invoca como punto de inicio del programa.


```python
#FUNCION 1-> Solicitud de datos
def solicitud_datos ():
  Altura= float(input ("Escribe tu ALTURA actual en metros:"))
  Peso= float(input ("Escribe tu PESO actual en kg:"))
  return Altura, Peso
  
#FUNCION 2 -> Calculo IMC
def calculo_IMC(Altura, Peso):
  IMC = Peso /(Altura**2)
  return IMC

#FUNCION 3 -> Validaciones
def validacion(Altura, Peso, IMC):
      if IMC < 18.5:
        categoría= "Bajo peso"
      elif IMC < 24.9:
        categoría="Peso normal"
      elif IMC < 29.9:
        categoría="Sobrepeso"
      elif IMC < 34.9:
        categoría="Obesidad grado I"
      elif IMC < 39.9:
        categoría="Obesidad grado II"
      else:
        categoría="Obesidad grado III"
      return categoría

# FUNCION GLOBAL -> MAIN
def main():
  Altura, Peso = solicitud_datos ()
  IMC = calculo_IMC(Altura, Peso)
  categoría= validacion(Altura, Peso, IMC)
  print (f"Tu IMC es: {IMC:.2f} y equivale a la categoría: {categoría}")
 

main ()
```

Esta función principal agrupa las tres funciones previamente descritas y organiza el flujo del código de manera secuencial. 🔴 EJERCICIO 3 - **Contador de palabras de un texto.**

Escribe una función que dado un texto, muestre las veces que aparece cada palabra.

Para realizar este ejercicio, cree las siguientes funciones que las uní luego en una única funcion llamada "main". A continuacion detallo las funciones: 

#FUNCION 1-> Solicitud de datos
La primera función que implementé se llama "solicitud_datos", y su propósito es solicitar al usuario que ingrese un texto. Para ello, utilizo el comando input, el cual presenta un mensaje claro para guiar al usuario sobre el tipo de datos que debe ingresar.
Una vez que el texto es ingresado, se aplica el método lower() a la entrada para convertir todas las letras a minúsculas, asegurando que el conteo de palabras sea insensible a mayúsculas y minúsculas.Finalmente, la función devuelve el texto procesado usando return, lo que permite que este valor sea utilizado en otras partes del código.

#FUNCION 2-> Ajustes del texto
Esta función toma como parámetro de entrada el texto proporcionado por la función "solicitud_datos". Su objetivo principal es realizar dos tareas:
1. Reemplazar caracteres especiales por espacios:para cumplir con este objetivo, primero determino cuáles son los caracteres especiales (,.!?;:()[]\"'-—/@) que deben ser reemplazados. Luego, utilizo un bucle for que recorre cada uno de esos caracteres y, con el método replace(), reemplaza cada uno de ellos por un espacio en blanco. Este proceso asegura que solo las palabras sean consideradas en el texto, eliminando cualquier signo de puntuación que pueda interferir con el análisis.
2. Dividir el texto en palabras: después de reemplazar los caracteres especiales, la siguiente tarea es dividir el texto en palabras. Para ello, aplico el método split(), que separa el texto en una lista de palabras basándose en los espacios entre ellas.
Finalmente, la función devuelve la lista de palabras utilizando return, lo que permite que este conjunto de palabras sea procesado por otras partes del código.

#FUNCION 3-> Contador de palabras
Esta función recibe como parámetro de entrada las palabras proporcionadas por la función "ajustes". En primer lugar, se crea un diccionario vacío llamado "conteo", que se utilizará para almacenar la frecuencia de cada palabra. Luego, mediante un bucle for, se recorre cada palabra en la lista de entrada. Dentro del bucle, se evalúa si la palabra ya existe como clave en el diccionario conteo. Si la palabra ya ha sido contada previamente, su valor en el diccionario se incrementa en 1, lo que indica un aumento en su frecuencia. Si la palabra no se encuentra en el diccionario, se agrega con un valor inicial de 1, ya que es la primera vez que aparece. Finalmente, la función retorna el diccionario conteo con el conteo actualizado de las palabras.

#FUNCION 4-> MAIN
Esta función agrupa las tres funciones previamente descritas, integrándolas en un único flujo de trabajo para simplificar su ejecución. Al unificar las funciones de solicitud de datos, ajustes y conteo de palabras, la función MAIN actúa como el punto central. Primero, solicita al usuario que ingrese el texto, luego aplica los ajustes necesarios a las palabras, y finalmente utiliza la función de conteo de palabras para calcular la frecuencia de cada término en el texto. Una vez procesado el conteo, la función muestra la frase ingresada junto con el conteo de cada palabra presente en el texto. Esta función MAIN se invoca para procesar el codigo. 

```python
#FUNCION 1-> Solicitud de datos
def solicitud_datos ():
  texto= input ("Escribe un texto" ).lower ()
  return texto

#FUNCION 2 -> Ajustes del texto (Reemplazo de caracteres especiales por espacios y dividir de texto en palabras)
def ajustes (texto):
  signos_puntuacion = ",.!?;:()[]\"'-—/@"
  for caracter in signos_puntuacion:
        texto = texto.replace(caracter, "") #Reemplaza signos por espacios
  palabras = texto.split() # division de palabras
  return palabras 

#FUNCION 3 -> Contador de palabras
def contador_palabras(palabras):
    conteo = {}  
    for palabra in palabras:
        if palabra in conteo:
            conteo[palabra] += 1
        else:
            conteo[palabra] = 1
    return conteo

#FUNCION 4 -> MAIN
def main ():
    texto=solicitud_datos ()
    palabras=ajustes (texto)
    conteo=contador_palabras(palabras)
    print (f"Este es el texto introducido: {texto}")
    print (f"Número de veces que se repite cada palabra en el texto: {conteo}")

main ()
    
```

🔴 EJERCICIO 4 - **Diccionario inverso.**

Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita a la inversa, es decir, intercambiados las claves y valores. Los valores y claves en el diccionario original son únicos; si éste no es el caso, la función debería imprimir un mensaje de advertencia.

reverse_dictionary({'a': 1, 'b': 2, 'c': 3})

{1: 'a', 2: 'b', 3: 'c'}

reverse_dictionary({'x': 'apple', 'y': 'banana', 'z': 'banana'})

Error: multiple keys for one value

Para realizar este ejercicio, cree las siguientes funciones que las uní luego en una única funcion llamada "main". A continuacion detallo las funciones: 

#FUNCION 1-> Definición de diccionarios
La primera función que implementé se llama "diccionarios", y su propósito es inicializar dos diccionarios con datos predefinidos. Estos diccionarios se asignan a las variables diccionario_1 y diccionario_2, respectivamente. Los datos dentro de estos diccionarios son proporcionados de acuerdo con el enunciado del ejercicio, y la función los retorna para su posterior uso. El primer diccionario contiene claves de tipo letra y valores numéricos, mientras que el segundo diccionario tiene claves de letras y valores de frutas. Estos diccionarios serán procesados por funciones posteriores para realizar las operaciones necesarias.

#FUNCION 2 -> Validaciones
Esta función recibe un diccionario como parámetro y verifica si los valores de las claves en ese diccionario son únicos. Compara la longitud de los valores en el diccionario con la longitud del conjunto de esos mismos valores. Si ambas longitudes coinciden, significa que no hay valores duplicados, ya que un conjunto no permite elementos repetidos. Para realizar la comparación, se utiliza la función len() aplicada tanto a diccionario.values() (que devuelve todos los valores del diccionario) como a set(diccionario.values()) (que elimina los duplicados al convertir los valores en un conjunto). La función retorna True si las longitudes coinciden (es decir, si los valores son únicos) y False en caso contrario.

#FUNCION 3 -> Diccionario inverso
La función "reverse_dictionary" recibe un diccionario como parámetro y verifica si sus valores son únicos mediante la función valores_unicos (Funcion 2). Si se encuentran valores duplicados, imprime un mensaje de error y retorna "None". Si los valores son únicos, invierte las claves y valores del diccionario, creando un nuevo diccionario donde las claves originales se convierten en valores y los valores originales se convierten en claves, utilizando una comprensión de diccionario ({valor: clave for clave, valor in diccionario.items()}). El resultado es un diccionario "invertido" si no hay duplicados en los valores.

#FUNCION 4 -> Procesar diccionarios
La función "procesar_diccionarios" recibe un número variable (*) de diccionarios como parámetros mediante el uso de *diccionarios, lo que permite pasar uno o más diccionarios a la función. Luego, recorre cada diccionario y llama a la función "reverse_dictionary" para invertir sus claves y valores. Los resultados de cada diccionario procesado se almacenan en una lista llamada "resultados". Finalmente, la función retorna la lista resultados, que contiene los diccionarios invertidos, uno por cada diccionario pasado como argumento. Si algún diccionario tiene valores duplicados, la función reverse_dictionary imprimirá un mensaje de error y retornará un diccionario vacío, el cual también se agregará a la lista de resultados.

##FUNCION 5-> MAIN
Esta función agrupa las funciones previamente descritas, integrándolas en un único flujo de trabajo para simplificar su ejecución. Al unificar las funciones actúa como el punto central. Primero, llama a la función diccionarios para obtener dos diccionarios, diccionario_1 y diccionario_2. Luego, pasa estos diccionarios a la función procesar_diccionarios, que invierte las claves y los valores de cada uno. El resultado de la función procesar_diccionarios es una lista que contiene los diccionarios invertidos (o vacíos si se detectaron valores duplicados). Finalmente, la función main recorre la lista resultados e imprime cada diccionario invertido, mostrando los resultados en la consola.




```python
#FUNCION 1 -> Definición de diccionarios
def diccionarios():
    diccionario_1 = {'a': 1, 'b': 2, 'c': 3}
    diccionario_2 = {'x': 'apple', 'y': 'banana', 'z': 'banana'}
    return diccionario_1, diccionario_2

#FUNCION 2 -> Validación de valores únicos
def valores_unicos(diccionario):
    return len(diccionario.values()) == len(set(diccionario.values()))

#FUNCION 3 -> Diccionario inverso
def reverse_dictionary(diccionario):
    if not valores_unicos(diccionario):
        return None, "Error: múltiples claves para un mismo valor"       
    return {valor: clave for clave, valor in diccionario.items()}
    
#FUNCION 4 -> Procesar diccionarios (version escalable)
def procesar_diccionarios(*diccionarios):
    resultados = []
    for diccionario in diccionarios:
        resultado = reverse_dictionary(diccionario)
        resultados.append(resultado)
    return resultados

#FUNCION 5-> MAIN
def main():
    diccionario_1, diccionario_2 = diccionarios()
    resultados = procesar_diccionarios(diccionario_1, diccionario_2) #paso directamente los diccionarios que tengo
    for resultado in resultados:
        print(resultado)


main()
```
